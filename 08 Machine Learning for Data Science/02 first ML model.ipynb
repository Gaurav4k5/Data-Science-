```python
from sklearn.tree import DecisionTreeClassifier
```

```python
feature = [[150,0],
           [170,0],
           [130,1],
           [120,1]]
```

```python
labels = ["apple", "apple", "orange", "orange"]
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(feature , labels)
```

```python
prediction = clf.predict([[90,0]])
```

```python
print(f"The fruit is {prediction[0]}")
```
